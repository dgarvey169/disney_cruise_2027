## 3. Budget & Cost Modeling <a id="budget"></a>

```python
# Budget Calculator
budget_items = {
    "Stateroom Base Fare": 5200.00,
    "Taxes, Fees & Port Expenses": 480.00,
    "Pre-Paid Gratuities ($14.50/person/night)": 14.50 * CRUISE_CONFIG["party_size"] * CRUISE_CONFIG["num_nights"],
    "Excursions / Port Adventures": 800.00,
    "Specialty Dining / Adult Lounges (Palo / Enchanté)": 250.00,
    "Travel & Flights": 1200.00,
    "Pre-Cruise Hotel (1 Night)": 250.00,
    "Onboard Souvenirs & Photo Package": 350.00,
}

df_budget = pd.DataFrame(list(budget_items.items()), columns=["Category", "Estimated_Cost_USD"])
total_estimated_cost = df_budget["Estimated_Cost_USD"].sum()
cost_per_person = total_estimated_cost / CRUISE_CONFIG["party_size"]

print(df_budget.to_string(index=False))
print("-" * 50)
print(f"💵 Total Estimated Cost: ${total_estimated_cost:,.2f}")
print(f"👤 Cost Per Person:      ${cost_per_person:,.2f}")
```

## 4. Booking Windows & Castaway Club Timeline <a id="timeline"></a>

DCL booking windows open based on your **Castaway Club** loyalty tier:
- **Pearl**: 123 days prior to sailing (Activities & Port Adventures)
- **Platinum**: 120 days prior
- **Gold**: 105 days prior
- **Silver**: 90 days prior
- **First-Time Guests**: 75 days prior
- **Online Check-in / Port Arrival Time (PAT)**: 30 days prior
